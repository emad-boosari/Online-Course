# Multi-qubit Gates

```python
import numpy as np

from qiskit import *
from qiskit_aer import *
from qiskit.quantum_info import *
from qiskit.visualization import *
```

## Visualization Function

```python
def qplot(x):
    return x.draw(fold = -1, initial_state = True, scale = 0.8, cregbundle = False)
```

## Multi-qubit Gates
1. CX and its Variations
2. CZ
3. CP
4. SWAP
5. CCX
6. CSWAP-Fredkin gate
7. multi-control X gates
   

## Variation of CNOT or CX Gate

```python
# create the multi-qubit register and classical register
q = QuantumRegister(2, name = 'qubit')
c = ClassicalRegister(2, name = 'bit')

# create the quantum circuit
qc = QuantumCircuit(q,c)
qc.x(q[0])
qc.barrier()
# Normal CX
qc.cx(q[0], q[1])
qc.barrier(label = 'Normal CX' )

# reverse CX
qc.cx(q[1], q[0])
qc.barrier(label = 'reverese CX')

# CX with open control
qc.x(q[0])
qc.cx(q[0], q[1])
qc.x(q[0])
qc.barrier(label = 'Open CX')

qc.x(q[1])
qc.cx(q[1], q[0])
qc.x(q[1])

qc.barrier(label = 'reverese open CX')


display(qplot(qc), Statevector(qc).draw('latex'))
```

## CZ, CP Gates

```python
qc = QuantumCircuit(2,2)
qc.x(0)
qc.x(1)
qc.barrier()

qc.cz(0,1)
qc.barrier(label = 'CZ')


qc.cp(np.pi/2, 0,1)


display(qplot(qc), Statevector(qc).draw('latex'))
```

## SWAP Gate

```python
qc = QuantumCircuit(2)
qc.x(0)

qc.swap(0,1)
display(qplot(qc), Statevector(qc).draw('latex'))
```

## CCX or Toffoli Gate

```python
qc = QuantumCircuit(3)
#qc.x(range(2))
qc.x(0)
qc.x(1)
qc.barrier()

qc.ccx(0,1,2)

display(qplot(qc), Statevector(qc).draw('latex'))
```

## CSWAP or Fredkin-Gate

```python
qc = QuantumCircuit(3)
qc.x(0)
qc.x(2)

qc.cswap(0,1,2)

display(qplot(qc), Statevector(qc).draw('latex'))
```

## Multi-contorl CNOT gate

```python
n = 5

qc = QuantumCircuit(n)
qc.x(range(4))
qc.mcx([0,1,2,3],4)

display(qplot(qc), Statevector(qc).draw('latex'))
```
