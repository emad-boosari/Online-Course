# Bell State Preparation

```python
from qiskit import *
from qiskit_aer import *
from qiskit.visualization import *
from qiskit.quantum_info import *

import numpy as np
```

```python
def qplot(x):
    return x.draw(fold=-1, initial_state= True, scale = 0.8)
```

## $|\Phi^+\rangle$
###### 

```python
qc1 = QuantumCircuit(2)
qc1.h(0)
qc1.cx(0,1)

display(qplot(qc1), Statevector(qc1).draw('latex'))
```

## $|\Phi_-\rangle$
###### 

```python
qc2 = QuantumCircuit(2)
qc2.h(0)
qc2.cx(0,1)
qc2.barrier()

qc2.z(1)

display(qplot(qc2), Statevector(qc2).draw('latex'))
```

## $|\Psi_+\rangle$
###### 

```python
qc3 = QuantumCircuit(2)
qc3.h(0)
qc3.cx(0,1)
qc3.barrier()

qc3.x(1)

display(qplot(qc3), Statevector(qc3).draw('latex'))
```

## $|\Psi_-\rangle$
###### 

```python
qc4 = QuantumCircuit(2)
qc4.h(0)
qc4.cx(0,1)
qc4.barrier()


qc4.x(1)
qc4.z(1)

display(qplot(qc4), Statevector(qc4).draw('latex'))
```

```python
circ = QuantumCircuit(2)
circ.ry(np.pi/2, 0)
circ.cx(0,1)

display(qplot(circ), Statevector(circ).draw('latex'))
```
